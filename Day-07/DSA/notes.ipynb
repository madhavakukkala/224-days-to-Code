## Quiz 1 — warm-up: predict the complexity

How many times does the loop body run? Time AND space complexity. Say it out loud.

```python
def quiz1(n):
    total = 0
    for i in range(n):
        total += i
    return total
```

**Answer:** O(n) time, O(1) space. One loop, runs n times, and only two variables (`total`, `i`) no matter how big n gets — like serving chai to n guests with the same one kettle.

## Quiz 2 — careful with this one

```python
def quiz2(n):
    count = 0
    while n > 0:
        count += 1
        n //= 10
    return count
```

**Answer:** O(log₁₀ n) time, O(1) space. The loop does NOT run n times — it runs once per **digit**. n = 1,00,000 has 6 digits, so just 6 rounds. Chopping a digit each round (`//= 10`) is the classic log pattern. Calling this O(n) is one of the most common interview mistakes.

## Quiz 3 — nested, but how nested?

```python
def quiz3(n):
    steps = 0
    for i in range(n):
        for j in range(i + 1):
            steps += 1
    return steps
```

**Answer:** O(n²) time, O(1) space. The inner loop runs 1, then 2, then 3 ... up to n times. Total = 1+2+...+n = n(n+1)/2. Half of n² is still n² — we drop the constant. Every triangle pattern you printed this week is exactly this shape.

## Quiz 4 — no loop visible... or is there?

```python
def quiz4(items, x):
    return x in items   # items is a normal Python list
```

**Answer:** O(n) time. `in` on a **list** checks every element one by one — the loop is hidden inside Python. On a **set** the same check is O(1) on average. One word (`list` vs `set`) changes the complexity — make sure this trap is on your cheat sheet.

## Quiz 5 — predict the OUTPUT (not complexity)

What exactly does this print? Both lines. Say it before running.

```python
a = [10, 20, 30]
b = a
b.append(40)
print(a)
print(b)
```

**Answer:** BOTH lines print `[10, 20, 30, 40]`. `b = a` did not copy the list — it stuck a second name-sticker on the SAME dabba. Change through either name, both see it. This is the Day-06 aliasing surprise. To get a real copy: `b = a.copy()` (shallow) or `copy.deepcopy(a)` for nested lists. If you predicted `a` unchanged — re-read Day-06 section 9 today.

## Quiz 6 — the trap that bites everyone once

What does this print? Think carefully about what `sort()` *returns*.

```python
marks = [67, 23, 90, 45]
result = marks.sort()
print(result)
print(marks)
```

**Answer:** `None`, then `[23, 45, 67, 90]`. `sort()` sorts the list **in place** and returns `None` — it hands you back nothing, like a tailor who alters your shirt but sends the receipt empty. If you want a sorted *photocopy* while keeping the original, use `sorted(marks)`. Writing `result = marks.sort()` and then using `result` is a classic silent bug.

## Bonus peek — tonight's topic, live

Tonight you learn what happens when you run a Python file. Here's a preview you can actually run: Python secretly compiles your function into **bytecode** — small instructions for the Python Virtual Machine. The `dis` module shows them.

```python
import dis

def add(a, b):
    return a + b

dis.dis(add)   # shows instructions like LOAD_FAST a, LOAD_FAST b, BINARY_OP +, RETURN_VALUE
```

Run it and see: your one line `return a + b` becomes a few tiny instructions — load `a`, load `b`, add, return. That instruction list is what actually executes, not your text. Full story in tonight's `dev/notes.md`.
