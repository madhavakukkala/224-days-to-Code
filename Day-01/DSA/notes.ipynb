# Day 01 — Python Basics (Interactive Notebook)

Welcome to your first notebook! A **notebook** is a file made of cells: some hold notes (like this one), some hold runnable code.

**How to use this notebook — the one rule:**
1. Read the code in a cell.
2. **Predict** the output. Say it out loud or write it down. Don't skip this!
3. Press `Shift + Enter` to run the cell.
4. Compare. If your guess was wrong, figure out *why* before moving on.

Predicting-then-running is how the concepts stick. Runners learn nothing; predictors learn everything.

## 1. Variables — labelled dabbas

A variable is a name pointing to a value, like a dabba labelled `sugar` with sugar inside. It is created the moment you assign to it with `=`.

**Predict:** what will the three prints show?

```python
# Creating variables — no announcement needed, just assign
city = "Jaipur"
platform_number = 4
ticket_price = 35.50

print(city)
print(platform_number)
print(ticket_price)
```

## 3. Dynamic typing — the dabba doesn't care what's inside

The same variable name can hold an int now and a string later. `type()` tells you what's inside *right now*.

**Predict:** the two `type()` outputs.

```python
score = 100
print(type(score))

score = "century"     # same name, new value, new type — Python allows it
print(type(score))
```

## 4. The 4 basic types

`int` (whole numbers), `float` (decimals), `str` (text in quotes), `bool` (`True`/`False`).

**Predict:** the four type names. Careful with the third one — quotes change everything!

```python
wickets = 3            # int
run_rate = 7.25        # float
team = "India"         # str
match_won = True       # bool  (capital T!)

print(type(wickets), type(run_rate), type(team), type(match_won))

# The classic trap: number vs number-in-quotes
print(type(2024), "vs", type("2024"))
```

## 5. Multiple assignment + the one-line swap

**Predict:** the values of `chai` and `coffee` after the swap line.

```python
# Fill many dabbas in one line
samosa, kachori, jalebi = 15, 12, 30
print(samosa, kachori, jalebi)

# The famous Python swap — no third glass needed
chai, coffee = "cutting", "filter"
chai, coffee = coffee, chai
print(chai, coffee)
```

## 6. print() — sep and end

`sep` = what goes **between** items (default: space). `end` = what goes **after** the line (default: newline `\n`).

**Predict:** exactly what the output looks like — how many lines in total?

```python
print("15", "08", "1947", sep="-")          # a date, stitched with hyphens
print("Mumbai", "Pune", "Nashik", sep=" -> ")
print("Loading", end="...")
print("done")                                # does this start a new line or not?
```

## 7. input() — and the trap that catches everyone

`input()` pauses and waits for you to type. **It ALWAYS returns a string**, even if you type digits.

**Run this and type `25`.** Predict the type it prints — int or str?

```python
age = input("Enter your age: ")
print("You entered:", age)
print("Its type is:", type(age))   # surprise incoming (or not, if you read the notes!)
```

## 8. Type casting — int(), float(), str()

Casting converts between types. Watch `int(7.9)` — it **chops**, never rounds.

**Predict:** all five printed values.

```python
print(int("42") + 8)       # str -> int, then maths works
print(float("3.14"))       # str -> float
print(int(7.9))            # chops the decimal — NOT rounding!
print(float(5))            # int -> float
print(str(97) + " runs")   # int -> str, then + glues text

# Crash gallery — predict the error, then uncomment ONE line at a time:
# int("3.14")   # int() refuses decimal STRINGS
# int("ten")    # words are not numbers
# int("")       # empty text
```

## 9. The input-casting combo you'll use every day

**Run this and type `7`.** Predict the output before running.

```python
n = int(input("Enter a number: "))   # input gives "7", int() makes it 7
print(n * 2)                          # real maths, no crash
# Without int(), what would "7" * 2 give? Try it in a new cell if curious!
```

## 10. f-strings — expressions and :.2f

Put `f` before the quote; `{}` become windows into the string. Whole expressions work inside. `:.2f` shows exactly 2 decimals (rounded).

**Predict:** all three lines, including the exact rupee value on line 3.

```python
player = "Kohli"
runs = 82
balls = 53

print(f"{player} scored {runs} off {balls} balls")
print(f"Fifty crossed? {runs >= 50}")               # comparison inside braces!
print(f"Strike rate: {runs / balls * 100:.2f}")     # calculation + 2 decimals
```

## 11. Operators — and the `/` float surprise

**Predict:** each printed value. Pay attention to `10 / 5` — int or float?

```python
print(7 + 3, 7 - 3, 7 * 3)
print(7 / 2)      # division: always float
print(10 / 5)     # even when it divides cleanly!
print(7 // 2)     # floor division: chops the decimal
print(7 % 2)      # modulo: the remainder
print(2 ** 10)    # power: 2 multiplied by itself 10 times
```

## 12. The digit-surgery twins: `% 10` and `// 10`

`n % 10` **reads** the last digit. `n // 10` **deletes** it. Together they munch a number right to left — the engine behind ALL five practice problems.

**Predict:** the full table this prints for 1947.

```python
n = 1947
print("n     | last digit | n after deleting it")
print(n,    "|", n % 10, "        |", n // 10)
n = n // 10
print(n,    " |", n % 10, "        |", n // 10)
n = n // 10
print(n,    "  |", n % 10, "        |", n // 10)
n = n // 10
print(n,    "   |", n % 10, "        |", n // 10)
# When n reaches 0 — the number is fully eaten. That's your stop signal.
```

## 🧩 Problem 1: Sum of first N numbers

**Goal:** N = 5 → 1+2+3+4+5 = 15.

**Hints:**
- A running total must start at 0 *before* any adding happens — like a fresh khata page each morning.
- The genius shortcut: pair first and last (1+5, 2+4...) → formula `N * (N + 1) // 2`. Why `//` instead of `/` here? (What type does `/` always return?)
- Check: N=1 → 1, N=10 → 55.

## 🧩 Problem 2: Reverse a number

**Goal:** 1234 → 4321.

**Hints:**
- Grab the last digit with `% 10`; delete it with `// 10`; stop when the number hits 0.
- Build the answer with `rev = rev * 10 + digit` — multiplying by 10 shifts digits left to make room (4 becomes 40, then +3 → 43).
- Test with 1200. You'll get 21, not 0021 — think about why that's actually correct.

## 🧩 Problem 3: Count digits

**Goal:** 1947 → 4.

**Hints:**
- Same munching loop, but you don't even need `%` — just chop with `// 10` and add 1 to a counter each time.
- Trap: what should the answer be for n = 0? What does your loop actually say? 🤔
- Sanity check: any number from 100 to 999 must give 3.

## 🧩 Problem 4: Palindrome number

**Goal:** 121 → yes; 123 → no. Reads the same both ways, like the name *Anna*.

**Hints:**
- You already built the reverse machine in Problem 2 — reuse it, then compare with `==`.
- Reversing destroys your working copy (it gets munched to 0). Save the original in a separate variable FIRST.
- Quick tests: 7 → yes (single digits always are), 10 → no, 1331 → yes.

## 🧩 Problem 5: Armstrong number

**Goal:** 153 → 1³ + 5³ + 3³ = 153 → yes! (3 digits, so power 3.)

**Hints:**
- Three phases: count the digits (Problem 3), then munch again raising each digit to that power (`**`, not `^`!) into a sum, then compare with the saved original.
- You pass over the number twice, so you need TWO fresh copies of it — the first pass eats one copy down to 0.
- The power = digit count. Don't hardcode 3 — 1634 is a 4-digit Armstrong number and should also pass.
- Test set: 153, 370, 371, 407 → yes. 123, 100 → no. Every single digit → yes.
