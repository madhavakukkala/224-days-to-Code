# Day 1 — Line-by-line walkthrough of main.py

Five problems solved today: sum of first N, reverse a number, count digits, palindrome number, Armstrong number.

Every function below is copied exactly from `main.py`. Each one gets three things: the idea, the code, and a line-by-line explanation with a small dry run.

Two operators run the whole show today:

- `num % 10` → the **last digit** of num (like taking the last coin off a stack)
- `num // 10` → num **with its last digit removed**

Repeat those two until the number hits 0, and you have dismantled it digit by digit.

## Problem 1 — Sum of first N numbers

**Idea:** add 1 + 2 + 3 + ... + n by keeping a running total, like a scoreboard adding runs ball by ball. For n = 10 the answer is 55.

```python
def Sum_of_first_N(self,n):
    sum = 0
    for i in range(1,n+1):
        sum += i
    return sum

# demo (passing None for self since we lifted the method out of the class)
print(Sum_of_first_N(None, 10))   # expected: 55
```

### Line by line

- `def Sum_of_first_N(self,n):` — in `main.py` this lives inside `class Solution()`, so it carries `self` (a reference to the object it belongs to). The logic only needs `n`.
- `sum = 0` — the running total starts empty. (Side note: `sum` is also the name of a built-in Python function; `total` would be a slightly safer name.)
- `for i in range(1,n+1):` — `range(1, n+1)` produces 1, 2, ..., n. The `+1` is crucial because `range` stops one **before** the end value.
- `sum += i` — shorthand for `sum = sum + i`; adds the current number to the total.
- `return sum` — hand back the final total.

### Dry run, n = 3

| i | sum after `sum += i` |
|---|---|
| 1 | 1 |
| 2 | 3 |
| 3 | 6 |

Returns 6. ✓ (1+2+3 = 6)

**Complexity:** O(n) time, O(1) space. Interview extra: the formula `n*(n+1)//2` gives the same answer in O(1) time.

## Problem 2 — Reverse a number

**Idea:** peel the last digit off the old number and push it onto the back of a new number. 453 becomes 354. The pushing trick is `reverse * 10 + digit` — multiplying by 10 shifts everything left to make room for the new digit.

```python
def reverse_a_number(self,number):
    reverse = 0
    while number > 0:
        digit = number % 10
        reverse = reverse *10 + digit
        number //= 10
    return reverse     

# demo
print(reverse_a_number(None, 453))   # expected: 354
```

### Line by line

- `reverse = 0` — the new number we are building, starts empty.
- `while number > 0:` — keep going as long as digits remain. When every digit has been peeled, `number` is 0 and the loop stops.
- `digit = number % 10` — grab the last digit.
- `reverse = reverse *10 + digit` — shift `reverse` left by one place, then drop the new digit into the ones place.
- `number //= 10` — chop the last digit off `number`. Same as `number = number // 10`.
- `return reverse` — the rebuilt, reversed number.

### Dry run, number = 453

| number | digit | reverse |
|---|---|---|
| 453 | 3 | 0·10 + 3 = 3 |
| 45 | 5 | 3·10 + 5 = 35 |
| 4 | 4 | 35·10 + 4 = 354 |
| 0 | — | loop ends |

Returns 354. ✓

**Complexity:** O(log n) time — the loop runs once per digit, and a number n has about log₁₀(n) digits. O(1) space.

## Problem 3 — Count digits

**Idea:** keep chopping the last digit off and count the chops. 1082945 needs 7 chops to reach 0, so it has 7 digits.

```python
def count_digits(self,num):
    count = 0
    while num > 0:
        num % 10
        count += 1
        num //=10
    return count

# demo
print(count_digits(None, 1082945))   # expected: 7
```

### Line by line

- `count = 0` — chop counter starts at zero.
- `while num > 0:` — loop while digits remain.
- `num % 10` — **did you notice?** This line computes the last digit... and throws it away. It is not assigned to anything, so Python evaluates it and discards the result. The function still works, because counting digits only needs the *number of chops*, not the digits themselves. This line does nothing and can simply be deleted. Good habit: every line should earn its place.
- `count += 1` — one more digit accounted for.
- `num //=10` — remove the last digit.
- `return count` — total digits.

### Dry run, num = 305

| num | count after this round |
|---|---|
| 305 | 1 |
| 30 | 2 |
| 3 | 3 |
| 0 | loop ends |

Returns 3. ✓

**Edge case:** `num = 0` returns 0, but zero actually has 1 digit. Worth handling if an interviewer asks.

**Complexity:** O(log n) time, O(1) space.

## Problem 4 — Palindrome number

**Idea:** a palindrome reads the same forwards and backwards, like MADAM or 32523. So: reverse the number (exact same loop as Problem 2) and check whether the reverse equals the original.

```python
def palindrome_number(self,num):
    number = num
    reverse = 0
    while number > 0:
        digit = number % 10
        reverse = reverse *10 + digit
        number //= 10
    return num == reverse

# demo
print(palindrome_number(None, 32523))   # expected: True
```

### Line by line

- `number = num` — **important copy step.** The loop below destroys its number digit by digit. We work on the copy (`number`) and keep the original (`num`) safe for the final comparison. Without this copy, we would end up comparing the reverse against 0.
- The middle four lines are exactly the reverse-a-number loop: peel digit, push into `reverse`, shrink `number`.
- `return num == reverse` — the comparison itself is the answer: `True` if palindrome, `False` if not. (A `True`/`False` value is called a **boolean**.)

### Dry run, num = 121

| number | digit | reverse |
|---|---|---|
| 121 | 1 | 1 |
| 12 | 2 | 12 |
| 1 | 1 | 121 |
| 0 | — | loop ends |

`num == reverse` → `121 == 121` → `True`. ✓ For 453, reverse is 354, so it returns `False`.

**Complexity:** O(log n) time, O(1) space.

## Problem 5 — Armstrong number

**Idea:** a number is an Armstrong number if the sum of each digit raised to the power of (total digit count) equals the number itself. 153 has 3 digits: 1³ + 5³ + 3³ = 1 + 125 + 27 = 153. Match!

Two passes: first count the digits (Problem 3 again), then rebuild the power-sum and compare.

```python
def armstrong_number(self,num):
    temp = num
    count = 0
    while temp > 0:
        temp % 10
        count += 1
        temp //=10


    temp = num
    sum = 0
    while temp>0:
        digit = temp % 10
        sum += digit ** count 
        temp //= 10

    return sum == num

# demo
print(armstrong_number(None, 153))   # expected: True
```

### Line by line

**First loop — count the digits:**

- `temp = num` — work on a copy so the original stays intact.
- `temp % 10` — **same "did you notice?" as Problem 3:** this bare line computes the last digit and discards it. Nothing breaks, because this loop only needs the count. The line can be removed.
- `count += 1` and `temp //=10` — count a digit, chop it off. After this loop, `count` = number of digits (3 for 153) and `temp` = 0.

**Second loop — build the power sum:**

- `temp = num` — refill `temp`, because the first loop ground it down to 0.
- `digit = temp % 10` — peel the last digit (this time we actually keep it).
- `sum += digit ** count` — `**` means "to the power of". Each digit is raised to `count` and added to the running sum.
- `temp //= 10` — chop and continue.

**Final:** `return sum == num` — `True` only if the power sum rebuilt the exact original number.

### Dry run, num = 153 (count = 3 from the first loop)

| temp | digit | digit ** 3 | sum |
|---|---|---|---|
| 153 | 3 | 27 | 27 |
| 15 | 5 | 125 | 152 |
| 1 | 1 | 1 | 153 |
| 0 | — | — | loop ends |

`sum == num` → `153 == 153` → `True`. ✓

**Complexity:** two passes over the digits is still O(log n) time. O(1) space.
