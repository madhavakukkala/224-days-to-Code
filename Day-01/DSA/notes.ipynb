## 1. Variables and types

A **variable** is a labelled dabba: the label (name) stays, the contents (value) can change.
Python decides the type from the value — no declaration needed.

**Predict before running:** the four `type()` lines should print `int`, `float`, `str`, `bool`. The last two lines show the *same label* pointing to a *new value* of a *different type* — perfectly legal in Python.

```python
runs = 74            # int  — whole number
price = 10.5         # float — has a decimal point
name = "Rohit"       # str  — text in quotes
is_out = False       # bool — only True or False

print(type(runs))    # <class 'int'>
print(type(price))   # <class 'float'>
print(type(name))    # <class 'str'>
print(type(is_out))  # <class 'bool'>

runs = "seventy four"   # same label, new contents, new type
print(type(runs))       # <class 'str'>
```

## 2. Multiple assignment and the swap trick

Python can fill many labelled dabbas in one line: `a, b = 1, 2`. And swapping two variables needs no third dabba — `a, b = b, a`. Python computes the whole right side first, then assigns. In most languages a swap needs a temp variable, like needing an empty plate to exchange two ladoos; Python exchanges them mid-air.

**Predict:** first print shows `1 2`, second shows `2 1` — swapped, no temp variable anywhere.

```python
a, b = 1, 2          # multiple assignment: a gets 1, b gets 2
print(a, b)          # 1 2

a, b = b, a          # the famous Pythonic swap — no third variable!
print(a, b)          # 2 1

runs, balls, fours = 82, 53, 9   # works for three (or more) at once
print(runs, balls, fours)
```

## 3. print() extras — sep and end

`print()` has two dials. `sep=` sets what goes *between* items (default: a space). `end=` sets what goes *after* the last item (default: a newline — that's why every print normally starts a fresh line).

**Predict:** line 1 separates with spaces, line 2 with commas, and the last two prints land on the SAME line.

```python
print("roti", "sabzi", "dal")              # roti sabzi dal — default sep is a space
print("roti", "sabzi", "dal", sep=", ")    # roti, sabzi, dal

print("Loading", end="")                   # end="" removes the newline
print("...done")                           # continues on the SAME line: Loading...done
```

## 4. input() and type casting

**`input()` always returns a string** — even if you type digits. `"25"` is text, 25 is a number; maths on text either crashes or does string-joining.

**Casting** converts between types: `int("25")` → 25, `str(74)` → `"74"`, `int(10.9)` → 10 (chops, never rounds), `float(7)` → 7.0 (always safe). Two casts CRASH: `int("abc")` and `int("10.5")` — they're left commented below so the cell still runs. The two-step rescue for `"10.5"`: `int(float("10.5"))`.

Note: `input()` may not work in a notebook preview (it waits for typing). The cell below *simulates* it with a hard-coded string so you can still see the trap.

**Predict:** first print shows `2525` (text repeated!), second shows `50` (real maths after casting).

```python
# Real programs use:  age = input("Enter age: ")
# We simulate the user typing 25:
age = "25"                 # input() gives a STRING, always

print(age * 2)             # "2525" — string repeated, not maths!
# print(age + 5)           # uncomment → TypeError: can't add str and int

age = int(age)             # cast: string → int
print(age * 2)             # 50 — now it is real maths

print(int(10.9))           # 10 — int() CHOPS the decimal, no rounding
print(float("10.5"))       # 10.5
print(float(7))            # 7.0 — int → float is always safe
print(str(74) + "0")       # "740" — number turned into text, then joined

# print(int("abc"))        # uncomment → ValueError: letters can't become a number
# print(int("10.5"))       # uncomment → ValueError: not a whole-number string
print(int(float("10.5"))) # 10 — the two-step route works
```

## 5. f-strings

An **f-string** has an `f` before the quote; anything inside `{ }` is replaced by its value — like filling blanks in a wedding card template.

**Predict:** line 1 fills in the blanks. Line 2 does maths *inside* the braces. Line 3 rounds to 2 decimal places (`:.2f`). Line 4 forgot the `f` — braces print literally. Spot the difference.

```python
name = "Virat"
runs = 82
bill = 100 / 3

print(f"{name} scored {runs} runs")          # Virat scored 82 runs
print(f"Ten more would be {runs + 10}")      # expressions allowed inside {}
print(f"Chai bill split 3 ways: Rs {bill:.2f}")  # Rs 33.33
print("{name} scored {runs} runs")           # forgot the f → braces print as-is
```

## 6. The two heroes: % and //

`a % b` (**modulo**) = remainder after dividing. Auto ride of Rs 47 paid in tens: 4 notes go in, Rs 7 is left over → `47 % 10 = 7`.

`a // b` (**floor division**) = divide and drop everything after the decimal.

The magic pair for digit problems:
- `num % 10` → the **last digit** (like taking the bottom coin off a stack)
- `num // 10` → the number **with its last digit removed**

**Predict:** `3`, then `45`, then `4.7` vs `4` (note: `/` gives a float, `//` gives an int), then `7` and `0` for a single-digit number.

```python
print(453 % 10)    # 3   — last digit peeled off
print(453 // 10)   # 45  — last digit chopped away

print(47 / 10)     # 4.7 — normal division: float
print(47 // 10)    # 4   — floor division: int, decimals dropped

print(7 % 10)      # 7   — a single digit IS its own last digit
print(7 // 10)     # 0   — chop the only digit and nothing remains
```

## 7. Peeling 453 step by step

Apply the pair three times by hand and watch the number shrink to 0. This is the whole trick, unrolled.

**Predict:** digits come out in order 3, 5, 4 (last to first), and the number goes 453 → 45 → 4 → 0.

```python
num = 453

print("digit:", num % 10)   # 3
num = num // 10
print("left :", num)        # 45

print("digit:", num % 10)   # 5
num = num // 10
print("left :", num)        # 4

print("digit:", num % 10)   # 4
num = num // 10
print("left :", num)        # 0  → nothing remains, we would stop here
```

## 8. The universal digit loop (building block, not a solution)

The manual peeling above, folded into a `while` loop. It just *visits* each digit and prints it. Every Day-1 problem is this skeleton plus one or two extra lines that you must figure out yourself.

`num //= 10` is shorthand for `num = num // 10`. Forget that line and the loop never ends!

**Predict:** it prints 3, 5, 4 (digits from last to first) and stops. Change 453 to 1082945 and re-run — you should see 7 digits.

```python
num = 453

while num > 0:          # keep going while digits remain
    digit = num % 10    # peel the last digit
    print(digit)        # <-- the problems replace this line with their own logic
    num //= 10          # chop it off; eventually num hits 0 and the loop stops
```

## 9. One last operator: ** (power) — and a / surprise

`a ** b` means "a to the power b". Needed for the Armstrong problem.

Full operator family: `+ - * / // % **`. The surprise member is `/` — it **always gives a float**, even when the division is exact.

**Predict:** 125, then 27, then 1 (1 to any power is 1), then `5.0` — yes, a float, even though 10 divides evenly by 2.

```python
print(5 ** 3)   # 125
print(3 ** 3)   # 27
print(1 ** 3)   # 1
# Fun check: 1 + 125 + 27 = 153 ... keep that number in mind for Armstrong.

print(10 / 2)   # 5.0 — / ALWAYS hands you a float; want an int? that's // 's job
```
