## 1. Indexing

A list is a train with numbered coaches. Index starts at 0. Negative index counts from the back: `-1` is the last coach.

**What to observe:** first item is `[0]` not `[1]`, and `[-1]` gives the last item without needing `len()`.

**Predicted output:**
```
45
90
23
5
```

````python
marks = [45, 67, 89, 23, 90]

print(marks[0])    # first coach
print(marks[-1])   # last coach
print(marks[-2])   # second last
print(len(marks))  # number of coaches (but last INDEX is len-1 = 4)
````

## 2. Slicing — `a[start:stop:step]`

**What to observe:** `stop` is EXCLUDED — `a[1:4]` gives indexes 1, 2, 3 only. Empty start means "from the beginning", empty stop means "till the end". Step `-1` walks backwards, so `a[::-1]` is a reversed copy.

**Predicted output:**
```
[20, 30, 40]
[10, 20, 30]
[30, 40, 50, 60]
[10, 30, 50]
[60, 50, 40, 30, 20, 10]
[10, 20, 30, 40, 50, 60]
```

Note the last line: even after `a[::-1]`, the original `a` is unchanged — slicing always makes a NEW list.

````python
a = [10, 20, 30, 40, 50, 60]

print(a[1:4])    # stop=4 is excluded
print(a[:3])     # from start
print(a[2:])     # till the end
print(a[::2])    # every 2nd item
print(a[::-1])   # reversed COPY
print(a)         # original untouched
````

## 3. Mutation — lists change in place, strings don't

**What to observe:** I can overwrite `a[0]` directly, and swap two items in one line without a temp variable. The same trick on a string throws `TypeError` (uncomment the last line to see it crash).

**Predicted output:**
```
[99, 2, 3]
[99, 3, 2]
[99, 3, 2, 7]
```

````python
a = [1, 2, 3]
a[0] = 99                 # overwrite in place
print(a)

a[1], a[2] = a[2], a[1]   # one-line swap — key trick for reversing!
print(a)

a.append(7)
print(a)

s = "abc"
# s[0] = "z"   # uncomment -> TypeError: strings are immutable
````

## 4. Method toolbox — sort vs sorted, extend vs append

**What to observe:** `a.sort()` changes `a` itself and returns `None` — so `b = a.sort()` is a classic bug. `sorted(a)` returns a new list and leaves the original alone. `extend` pours items in one by one; `append` drops the whole list in as ONE item.

**Predicted output:**
```
sort() returned: None
a after sort(): [1, 2, 3]
sorted copy: [1, 5, 9]  original: [9, 1, 5]
extend: [1, 2, 3, 4, 5]
append: [1, 2, 3, [4, 5]]
index of 2: 1  count of 3: 1
2 in a: True
```

````python
a = [3, 1, 2]
result = a.sort()          # sorts IN PLACE, returns None
print("sort() returned:", result)
print("a after sort():", a)

c = [9, 1, 5]
print("sorted copy:", sorted(c), " original:", c)   # c untouched

x = [1, 2, 3]
x.extend([4, 5])           # pour items in one by one
print("extend:", x)

y = [1, 2, 3]
y.append([4, 5])           # the whole list becomes ONE item!
print("append:", y)

print("index of 2:", a.index(2), " count of 3:", a.count(3))
print("2 in a:", 2 in a)   # membership check — walks the list, O(n)
````

## 5. Looping — enumerate beats range(len)

**What to observe:** `enumerate` hands me index AND value together — no `items[i]` typing, no off-by-one risk. `zip` walks two lists side by side.

**Predicted output:**
```
0 idli 20
1 dosa 40
2 poha 25
idli costs 20
dosa costs 40
poha costs 25
```

````python
items = ["idli", "dosa", "poha"]
prices = [20, 40, 25]

# enumerate: index + value together (cleaner than range(len(items)))
for i, item in enumerate(items):
    print(i, item, prices[i])

# zip: two lists walking side by side
for item, price in zip(items, prices):
    print(item, "costs", price)
````

## 6. The aliasing surprise — `b = a` is NOT a copy

One tiffin box, two names. `b = a` gives the SAME list a second name. Append through `b`, and `a` "changes" too — because there is only one list.

**What to observe:** printing `a` shows the 4 even though I appended via `b`. Also `a is b` is `True` (same object in memory).

**Predicted output:**
```
b = [1, 2, 3, 4]
a = [1, 2, 3, 4]
a is b: True
```

````python
a = [1, 2, 3]
b = a            # NOT a copy — just a second name for the same list
b.append(4)

print("b =", b)
print("a =", a)          # surprise! a also shows 4
print("a is b:", a is b) # 'is' checks: same object?
````

## 7. Shallow copy — new outer box, SAME inner boxes

`a.copy()` (or `a[:]`) fixes the aliasing problem for flat lists. But if the list contains inner lists, only the OUTER list is copied — the inner lists are still shared. New tiffin carrier, same old small dabbas inside.

**What to observe:** appending to the copy's outer level does NOT touch `a`, but appending inside `b[0]` DOES leak into `a`.

**Predicted output:**
```
after outer append: a = [[1, 2], [3, 4]]
after inner append: a = [[1, 2, 99], [3, 4]]
same inner list? True
```

````python
a = [[1, 2], [3, 4]]
b = a.copy()               # shallow copy

b.append([5, 6])           # outer-level change: a is safe
print("after outer append: a =", a)

b[0].append(99)            # inner-level change: LEAKS into a!
print("after inner append: a =", a)

print("same inner list?", a[0] is b[0])
````

## 8. Deep copy — fully independent duplicate

`copy.deepcopy(a)` clones the outer list AND every inner list, all the way down. Nothing is shared.

**What to observe:** the same inner append that leaked in the previous cell now leaves `a` untouched, and `a[0] is b[0]` is `False`.

**Predicted output:**
```
a = [[1, 2], [3, 4]]
b = [[1, 2, 99], [3, 4]]
same inner list? False
```

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)       # fully independent clone

b[0].append(99)            # no leak this time
print("a =", a)
print("b =", b)
print("same inner list?", a[0] is b[0])
````

## 9. List comprehensions — one-line list builders

Pattern: `[expression for item in iterable if condition]`. Read left to right like English.

**What to observe:** the comprehension gives exactly the same result as the 3-line loop, and the `if` at the end filters items out.

**Predicted output:**
```
loop way:          [0, 1, 4, 9, 16]
comprehension way: [0, 1, 4, 9, 16]
evens: [0, 2, 4, 6, 8]
long words: ['VADA PAV']
```

````python
# the old way
squares = []
for x in range(5):
    squares.append(x * x)
print("loop way:         ", squares)

# the one-line way
squares2 = [x * x for x in range(5)]
print("comprehension way:", squares2)

# with an if filter
evens = [x for x in range(10) if x % 2 == 0]
print("evens:", evens)

words = ["chai", "vada pav", "dosa"]
print("long words:", [w.upper() for w in words if len(w) > 4])
````

## 10. Comprehensions part 2 — if-else inside, nested

**What to observe:** an `if` at the END filters (some items get dropped). An `if-else` at the FRONT transforms (every item kept, in one of two forms). In the nested version, the left loop is the OUTER loop.

**Predicted output:**
```
filter: [80, 50]
transform: ['fail', 'pass', 'pass']
pairs: [(1, 'a'), (1, 'b'), (2, 'a'), (2, 'b')]
```

````python
marks = [30, 80, 50]

# if at the END = filter (keeps only some items)
passed = [m for m in marks if m >= 35]
print("filter:", passed)

# if-else at the FRONT = transform (keeps ALL items)
result = ["pass" if m >= 35 else "fail" for m in marks]
print("transform:", result)

# nested: loop inside a loop (the left one is the OUTER loop)
pairs = [(x, y) for x in [1, 2] for y in ["a", "b"]]
print("pairs:", pairs)
````

### 1. Reverse an array in place
- No new list allowed — so `a[::-1]` is banned here.
- Two pointers: `left = 0`, `right = len(a) - 1`. While `left < right`: swap `a[left], a[right] = a[right], a[left]`, then move both inward.
- Picture two friends entering a train from opposite ends, exchanging seats at each step until they meet.
- Test with even length, odd length, and a single-element list.

### 2. Find min and max
- One pass. Start `mn = mx = a[0]` — NOT 0 (all-negative lists will break a 0 start).
- For each element: smaller than `mn`? update. Bigger than `mx`? update.
- Don't sort — that is O(n log n) for an O(n) job. Decide what to do for an empty list.

### 3. Second largest WITHOUT sorting
- Track TWO values in one pass: `largest` and `second`.
- When `x > largest`: FIRST `second = largest`, THEN `largest = x`. Order matters — swap the order and the old champion is lost.
- Else if `x > second` and `x != largest`: update `second`.
- Duplicate gotcha: test `[10, 10, 5]` and `[5, 10, 10]`. Also decide what to return for `[7]` or `[7, 7]` (no second largest exists).

### 4. Left rotate by one
- `[1, 2, 3, 4, 5]` becomes `[2, 3, 4, 5, 1]`.
- Save `first = a[0]` BEFORE shifting, or it gets overwritten and lost.
- Shift: for `i` from 1 to end, `a[i - 1] = a[i]`. Then `a[-1] = first`.
- Handle empty and single-element lists (nothing to do).
