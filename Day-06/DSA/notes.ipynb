## 6. Shallow vs deep — the nested trap

We copied `students`... or did we? Predict what happens to the ORIGINAL after we edit the copy.

```python
students = [["Ravi", 72], ["Priya", 85]]
shallow = students.copy()
shallow[0][1] = 99
print("original:", students)
```

```python
import copy

students = [["Ravi", 72], ["Priya", 85]]
deep = copy.deepcopy(students)
deep[0][1] = 40
print("original:", students)
print("deep copy:", deep)
```

### Related trap: building a grid with *

Predict the printed grid after changing ONE cell.

```python
grid = [[0]] * 3
grid[0][0] = 9
print(grid)

safe = [[0] for _ in range(3)]
safe[0][0] = 9
print(safe)
```
