## 1. if / elif / else — auto-meter fare slabs

**Observe:** Python checks conditions top to bottom and runs only the FIRST block whose condition is true. `km = 4.0` fails the first check (`<= 1.5`), passes the second (`<= 5`), so only the `elif` block runs and `else` is never even looked at.

**Predicted output:** `Fare: 62.5` — because 25 + (4.0 − 1.5) × 15 = 25 + 37.5.

Try after running: change `km` to 1.0 (should hit the flat slab, ₹25) and to 8.0 (should hit `else`).

````python
km = 4.0

if km <= 1.5:
    fare = 25                                  # flat minimum fare
elif km <= 5:
    fare = 25 + (km - 1.5) * 15                # ₹15 per extra km
else:
    fare = 25 + 3.5 * 15 + (km - 5) * 18       # longer trips cost more per km

print("Fare:", fare)
````

### Order of conditions matters

**Observe:** the SAME slabs, but the wide check (`<= 5`) is placed first. A 1 km ride is also ≤ 5, so it gets caught by the wrong slab. First true condition wins — always put the narrowest slab on top.

**Predicted output:** `Wrong order fare: 17.5` (it charged the per-km slab), then `Right order fare: 25`.

````python
km = 1.0

# WRONG: wide condition first swallows the small ride
if km <= 5:
    fare = 25 + (km - 1.5) * 15
elif km <= 1.5:
    fare = 25
print("Wrong order fare:", fare)

# RIGHT: narrow condition first
if km <= 1.5:
    fare = 25
elif km <= 5:
    fare = 25 + (km - 1.5) * 15
print("Right order fare:", fare)
````

## 2. range() variations — stop is EXCLUDED

**Observe:** wrapping a range in `list(...)` shows every number it would produce. Key things to catch:

- `range(5)` starts at 0 and **5 itself is missing** — stop is excluded, like a train that halts just before the named station.
- To actually reach `n`, you need `range(1, n + 1)`.
- Negative step counts DOWN, and then start must be bigger than stop.
- `range(5, 0)` with no step is empty — default step +1 cannot climb from 5 to 0.

**Predicted output:**
```
[0, 1, 2, 3, 4]
[2, 3, 4, 5]
[1, 3, 5, 7, 9]
[5, 4, 3, 2, 1]
[]
```

````python
print(list(range(5)))          # 0..4, five numbers, 5 excluded
print(list(range(2, 6)))       # 2..5, six excluded
print(list(range(1, 10, 2)))   # odd numbers, step 2
print(list(range(5, 0, -1)))   # countdown 5..1
print(list(range(5, 0)))       # EMPTY: default step +1 cannot go down
````

## 3. while with a counter — filling a bucket

**Observe:** we do not know beforehand how many mugs it takes; we only know the stopping condition (bucket full). The condition is checked BEFORE every round. The line `bucket = bucket + 3` is the loop's escape route — delete it and the loop never ends (infinite loop).

**Predicted output:**
```
Mug 1 poured, bucket = 3
Mug 2 poured, bucket = 6
Mug 3 poured, bucket = 9
Mug 4 poured, bucket = 12
Bucket full after 4 mugs
```
Note it overshoots to 12: the check `9 < 10` was still true, so a 4th mug was poured.

````python
bucket = 0        # litres in the bucket right now
capacity = 10
mugs = 0          # counter: how many mugs poured

while bucket < capacity:
    bucket = bucket + 3      # one mug = 3 litres  <-- the escape route
    mugs = mugs + 1
    print("Mug", mugs, "poured, bucket =", bucket)

print("Bucket full after", mugs, "mugs")
````

## 4a. break — batsman out, innings over

**Observe:** the loop was set up for 6 balls, but `break` on ball 4 kills it instantly. Balls 5 and 6 never happen; control jumps to the line after the loop.

**Predicted output:**
```
Ball 1 : run taken
Ball 2 : run taken
Ball 3 : run taken
OUT on ball 4 — innings over
Next batsman walks in
```

````python
for ball in range(1, 7):        # an over: balls 1..6
    if ball == 4:
        print("OUT on ball", ball, "— innings over")
        break                   # exit the WHOLE loop right now
    print("Ball", ball, ": run taken")

print("Next batsman walks in")
````

## 4b. continue — dot ball, move to next delivery

**Observe:** `continue` skips only the REST of the current round, then the loop carries on with the next value. Ball 3 produces no print, but balls 4, 5, 6 still happen — compare with `break` above, which ended everything.

**Predicted output:**
```
Ball 1 : scored
Ball 2 : scored
Ball 4 : scored
Ball 5 : scored
Ball 6 : scored
Over complete
```

````python
for ball in range(1, 7):
    if ball == 3:
        continue                # dot ball: skip the print, next delivery
    print("Ball", ball, ": scored")

print("Over complete")
````

## 5. Nested loops — every (row, chair) coordinate

**Observe:** for every ONE step of the outer loop, the inner loop runs COMPLETELY. Watch the pattern in the output: `i` stays frozen while `j` sprints 1→4, then `i` ticks once and `j` sprints again. 3 outer steps × 4 inner steps = 12 lines.

This is shaadi seating: stand at row 1, greet chairs 1–4, only then move to row 2.

**Predicted output (first 5 lines):**
```
i=1 j=1
i=1 j=2
i=1 j=3
i=1 j=4
i=2 j=1
```
...and so on until `i=3 j=4`.

````python
for i in range(1, 4):        # 3 rows
    for j in range(1, 5):    # 4 chairs in each row
        print("i=" + str(i), "j=" + str(j))
    # inner loop fully done -> outer loop takes its next step
````

## 6. The 3-step pattern recipe — one full worked example

The recipe for EVERY pattern problem:

1. **Outer loop = rows.** Count rows in the picture → `for i in range(1, n + 1)`.
2. **Inner loop(s) = one row's contents as a formula in `i`.** Make a table: row number vs how many spaces / how many symbols. Read the formula off the table. If a row has spaces then symbols, use two inner loops back to back.
3. **Bare `print()` per row.** Inside a row print with `end=""` (stay on the same line); the empty `print()` after the inner loops jumps to the next row.

### Demo (my own, not one of the 8): right-ALIGNED triangle of `#`, n = 5

```
....#      <- dots mark where spaces go
...##
..###
.####
#####
```

**The table** (counted straight off the picture):

| row `i` | spaces | hashes |
|---|---|---|
| 1 | 4 | 1 |
| 2 | 3 | 2 |
| 3 | 2 | 3 |
| 4 | 1 | 4 |
| 5 | 0 | 5 |

Hashes = `i` directly. Spaces = 4,3,2,1,0 against `i` = 1..5 → `n - i`. **Edge check:** `i=1` → 4 spaces ✓, `i=5` → 0 spaces ✓. Both edges match, so the formulas are right.

**Observe in the code below:** the spaces loop finishes fully before the hash loop starts (two inner loops in sequence, not nested in each other), and the final `print()` lines up with the inner `for`s — once per row, not once per symbol.

**Predicted output:** the triangle above, with real spaces instead of dots.

````python
n = 5
for i in range(1, n + 1):        # step 1: rows, i = 1..5
    for s in range(n - i):       # step 2a: spaces first (n - i of them)
        print(" ", end="")
    for h in range(i):           # step 2b: then i hashes
        print("#", end="")
    print()                      # step 3: row finished, new line
````
