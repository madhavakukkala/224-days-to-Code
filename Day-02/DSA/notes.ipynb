# Day 2 — Pattern Problems 1 to 8 (Line-by-Line Walkthrough)

These are my solutions from `main.py`, explained slowly with dry runs.

The recipe used everywhere:

1. Outer loop `i` = which row we are on.
2. Inner loop(s) `j` = what gets printed in that row (spaces / stars / numbers), as a formula in `i`.
3. `print("*", end="")` keeps things on the same line; the bare `print()` at the end moves to the next row.

All dry runs below use `n = 5`.

## Pattern 01 — Solid Rectangle

**Idea:** Every row looks the same: `n` stars. So both loops simply run `n` times. This is the "hello world" of patterns — like a full box of laddoos, every row packed edge to edge.

Expected output for `n = 5`:

```
*****
*****
*****
*****
*****
```

````python
def pattern01(self,n):
    for i in range(n):
        for j in range(n):
            print("*", end="")
        print()
````

### Line by line

- `for i in range(n):` — outer loop. `range(5)` gives `0, 1, 2, 3, 4`, so we get 5 rows. `i` is the row number (0-based).
- `for j in range(n):` — inner loop. For **each** row, this runs 5 times and prints 5 stars.
- `print("*", end="")` — print a star and **stay on the same line** (`end=""` replaces the default newline with nothing).
- `print()` — inner loop is done, the row is complete, jump to the next line. Note its indentation: it lines up with the inner `for`, so it runs once per row, not once per star.

### Dry run (n = 5)

| `i` | inner loop `j` values | stars printed | after `print()` |
|---|---|---|---|
| 0 | 0,1,2,3,4 | `*****` | newline |
| 1 | 0,1,2,3,4 | `*****` | newline |
| 2 | 0,1,2,3,4 | `*****` | newline |
| 3 | 0,1,2,3,4 | `*****` | newline |
| 4 | 0,1,2,3,4 | `*****` | newline |

Total prints: 5 × 5 = 25 stars. The inner loop does not care about `i` here — that is exactly why every row is identical.

## Pattern 02 — Right Triangle of Stars

**Idea:** Row `i` (0-based) should have `i + 1` stars: row 0 → 1 star, row 1 → 2 stars, and so on. So the inner loop runs `i + 1` times. The pattern grows like a school assembly line-up: shortest first, one more kid per row.

Expected output for `n = 5`:

```
*
**
***
****
*****
```

````python
def pattern02(self,n):
    for i in range(n):
        for j in range(i+1):
            print("*", end="")
        print()
````

### Line by line

- `for i in range(n):` — 5 rows, `i` goes `0 → 4`.
- `for j in range(i+1):` — the key line. The inner loop's size **depends on `i`**. `range(i+1)` gives `i + 1` values, so the star count grows with the row.
- `print("*", end="")` — one star, same line.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | `range(i+1)` | `j` values | stars |
|---|---|---|---|
| 0 | `range(1)` | 0 | `*` |
| 1 | `range(2)` | 0,1 | `**` |
| 2 | `range(3)` | 0,1,2 | `***` |
| 3 | `range(4)` | 0,1,2,3 | `****` |
| 4 | `range(5)` | 0,1,2,3,4 | `*****` |

Why `i+1` and not `i`? Because on the first row `i = 0`, and `range(0)` runs zero times — we would print an empty first row. `i + 1` fixes that off-by-one.

## Pattern 03 — Number Triangle (1, 12, 123, ...)

**Idea:** Same triangle shape as Pattern 02, but instead of stars, row `i` prints the numbers `1 2 3 ... i`. Trick: print the loop variable `j` itself. To make the maths clean, both loops start from 1 instead of 0.

Expected output for `n = 5`:

```
1
12
123
1234
12345
```

````python
def pattern03(self,n):
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(j, end="")
        print()
````

### Line by line

- `for i in range(1,n+1):` — rows numbered `1 → 5` (1-based now). Remember: stop is excluded, so `n+1` is needed to actually reach `n`.
- `for j in range(1,i+1):` — `j` walks `1, 2, ..., i`. Row 3 → `j` is 1, 2, 3.
- `print(j, end="")` — we print **`j` itself**, not a star. That is the whole difference from Pattern 02.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | `range(1,i+1)` | `j` values | printed |
|---|---|---|---|
| 1 | `range(1,2)` | 1 | `1` |
| 2 | `range(1,3)` | 1,2 | `12` |
| 3 | `range(1,4)` | 1,2,3 | `123` |
| 4 | `range(1,5)` | 1,2,3,4 | `1234` |
| 5 | `range(1,6)` | 1,2,3,4,5 | `12345` |

Starting both loops at 1 avoids printing a 0 — with 0-based loops we would get `0`, `01`, `012`...

## Pattern 04 — Number Triangle (1, 22, 333, ...)

**Idea:** Same shape again, but now row `i` prints the **row number itself**, `i` times. One-character change from Pattern 03: print `i` instead of `j`. The inner loop still controls *how many* times; it just no longer decides *what* is printed.

Expected output for `n = 5`:

```
1
22
333
4444
55555
```

````python
def pattern04(self,n):
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(i, end="")
        print()
````

### Line by line

- `for i in range(1,n+1):` — rows `1 → 5`, 1-based.
- `for j in range(1,i+1):` — runs `i` times. Here `j` is only a counter; its value is never used.
- `print(i, end="")` — prints the **row number**. Row 4 prints `4`, four times.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | inner runs | what is printed each time | row output |
|---|---|---|---|
| 1 | 1 time | `1` | `1` |
| 2 | 2 times | `2` | `22` |
| 3 | 3 times | `3` | `333` |
| 4 | 4 times | `4` | `4444` |
| 5 | 5 times | `5` | `55555` |

Patterns 02, 03, 04 are the same skeleton. Only the thing inside `print(...)` changes: `"*"` → `j` → `i`. Realising this is the real lesson.

## Pattern 05 — Inverted Right Triangle

**Idea:** Pattern 02 flipped upside down. Row 1 has 5 stars, row 2 has 4... so row `i` (1-based) has `n - i + 1` stars. As `i` goes up, the star count goes down — like overs remaining in a chase: every over bowled, one less remains.

Expected output for `n = 5`:

```
*****
****
***
**
*
```

````python
def pattern05(self,n):
    for i in range(1,n+1):
        for j in range(n-i+1):
            print("*", end="")
        print()
````

### Line by line

- `for i in range(1,n+1):` — rows `1 → 5`.
- `for j in range(n-i+1):` — the formula. Check it at the edges: `i = 1` → `range(5)` → 5 stars. `i = 5` → `range(1)` → 1 star. Edges correct means the formula is correct.
- `print("*", end="")` — star, same line.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | `n-i+1` | stars |
|---|---|---|
| 1 | 5 | `*****` |
| 2 | 4 | `****` |
| 3 | 3 | `***` |
| 4 | 2 | `**` |
| 5 | 1 | `*` |

How to find such formulas: make this exact table on paper FIRST (row → count), then ask "what expression in `i` gives this column?" Here 5,4,3,2,1 against 1,2,3,4,5 → `6 - i` → `n - i + 1`.

## Pattern 06 — Inverted Number Triangle

**Idea:** Pattern 03 flipped. Row `i` prints `1 2 ... (n - i + 1)`. The inner loop must produce actual numbers starting at 1, so it is `range(1, n-i+2)` — start at 1, and the stop needs `+2` because stop is excluded.

Expected output for `n = 5`:

```
12345
1234
123
12
1
```

````python
def pattern06(self,n):
    for i in range(1,n+1):
        for j in range(1,n-i+2):
            print(j, end = "")
        print()
````

### Line by line

- `for i in range(1,n+1):` — rows `1 → 5`.
- `for j in range(1,n-i+2):` — we want `j` to reach `n - i + 1` (the last number of the row). Since `range` excludes the stop, we write `n - i + 2`. This "+1 more than what I want" adjustment is the single most common source of bugs in patterns.
- `print(j, end = "")` — print the number itself.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | `range(1, n-i+2)` | `j` values | printed |
|---|---|---|---|
| 1 | `range(1,6)` | 1,2,3,4,5 | `12345` |
| 2 | `range(1,5)` | 1,2,3,4 | `1234` |
| 3 | `range(1,4)` | 1,2,3 | `123` |
| 4 | `range(1,3)` | 1,2 | `12` |
| 5 | `range(1,2)` | 1 | `1` |

Sanity check trick: plug in the first and last row. `i=1` should give the longest row, `i=n` should give just `1`. Both work, so the formula is right.

## Pattern 07 — Pyramid

**Idea:** First pattern with TWO inner loops per row: one for spaces, one for stars. To centre the pyramid, each row prints some spaces first, then an **odd** number of stars (1, 3, 5, 7, 9). For 0-based row `i`: spaces = `n - i - 1`, stars = `2*i + 1`. Spaces shrink as stars grow, so everything stays centred — like Diwali diyas arranged into a triangle on the floor.

Expected output for `n = 5`:

```
    *
   ***
  *****
 *******
*********
```

````python
def pattern07(self,n):
    for i in range(n):

        #spaces
        for j in range(n-i-1):
            print(" ", end = "")
        #stars
        for j in range(0,2*i+1):
            print("*", end = "")
        print()
````

### Line by line

- `for i in range(n):` — 5 rows, `i` goes `0 → 4` (back to 0-based).
- `for j in range(n-i-1):` — the **spaces** loop. Top row (`i=0`) needs the most pushing-in: `5-0-1 = 4` spaces. Bottom row (`i=4`): `0` spaces.
- `for j in range(0,2*i+1):` — the **stars** loop. `2*i + 1` gives the odd numbers 1, 3, 5, 7, 9. Odd counts are what keep a pyramid symmetric around its centre.
- Reusing `j` in both loops is fine — the first loop is completely finished before the second starts.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | spaces `n-i-1` | stars `2i+1` | row (spaces shown as `.`) |
|---|---|---|---|
| 0 | 4 | 1 | `....*` |
| 1 | 3 | 3 | `...***` |
| 2 | 2 | 5 | `..*****` |
| 3 | 1 | 7 | `.*******` |
| 4 | 0 | 9 | `*********` |

Check the width: spaces + stars = `(n-i-1) + (2i+1) = n + i`. Bottom row width = `2n - 1 = 9`. Also note: no spaces AFTER the stars — trailing spaces are invisible, so we never bother printing them.

## Pattern 08 — Inverted Pyramid

**Idea:** Pattern 07 upside down. Now spaces GROW with the row and stars SHRINK: for 0-based row `i`, spaces = `i` and stars = `2n - 2i - 1` (9, 7, 5, 3, 1).

Expected output for `n = 5`:

```
*********
 *******
  *****
   ***
    *
```

````python
def pattern08(self,n):
    for i in range(n):
        for j in range(i):
            print(" ", end = "")

        for j in range(2*n - 2*i - 1):
            print("*", end = "")
        print()
````

### Line by line

- `for i in range(n):` — 5 rows, `i` from `0 → 4`.
- `for j in range(i):` — spaces loop. Row 0 gets `range(0)` → **zero** spaces (the widest row sits flush left). Each next row is pushed in by one more space.
- `for j in range(2*n - 2*i - 1):` — stars loop. At `i=0`: `2*5 - 0 - 1 = 9` stars. Each row `i` increases by 1, so the count drops by 2. That is where the `- 2*i` comes from.
- `print()` — end of row.

### Dry run (n = 5)

| `i` | spaces `i` | stars `2n-2i-1` | row (spaces shown as `.`) |
|---|---|---|---|
| 0 | 0 | 9 | `*********` |
| 1 | 1 | 7 | `.*******` |
| 2 | 2 | 5 | `..*****` |
| 3 | 3 | 3 | `...***` |
| 4 | 4 | 1 | `....*` |

Compare with Pattern 07: the two formulas literally swap roles. Pyramid: spaces shrink (`n-i-1`), stars grow (`2i+1`). Inverted: spaces grow (`i`), stars shrink (`2n-2i-1`). Stack Pattern 07 on top of Pattern 08 and you get a diamond — that is exactly what `pattern09` does (next session).
