## 5. `and` / `or` / `not` — and the famous trap

**Predict:** x = 100. Does the second `if` print? Think hard before running.

````python
x = 100

if x == 5 or x == 6:          # correct way
    print("x is 5 or 6")

if x == 5 or 6:               # THE TRAP
    print("this prints EVERY time — why?")
````

Python reads `x == 5 or 6` as `(x == 5) or (6)`. The bare `6` is truthy, so the whole condition is always True. Each side of `or` must be a **complete question**.

## 9. `for` + `range()` — the number machine

**Predict:** write down exactly what each loop prints BEFORE running.

````python
print("range(5):")
for i in range(5):
    print(i, end=" ")        # end=" " from yesterday: no newline jump

print("\nrange(1, 6):")
for i in range(1, 6):
    print(i, end=" ")

print("\nrange(2, 11, 2):")
for i in range(2, 11, 2):
    print(i, end=" ")

print("\nrange(5, 0, -1):")
for i in range(5, 0, -1):
    print(i, end=" ")
````

**Predict:** what does this print? (Trick question.)

````python
for i in range(5, 0):    # counting down WITHOUT step -1
    print(i)
print("loop finished — did anything print above?")
````

`range(5, 0)` is an **empty range** — with the default step +1 you cannot walk from 5 to 0, so the body runs zero times. No error, just silence. Counting down needs `step=-1`.

## 10. Counter vs items — two looping styles

**Predict:** what does each loop print?

````python
menu = "dosa"

# Style 1: give me the ITEMS
for ch in menu:
    print(ch, end=" ")

print()

# Style 2: give me the POSITIONS (counter)
for i in range(len(menu)):
    print(i, menu[i])
````

## 12. `break`, `continue`, `pass`

**Predict:** what numbers does EACH loop print?

````python
print("break:")
for i in range(1, 11):
    if i == 4:
        break               # kill the whole loop
    print(i, end=" ")

print("\ncontinue:")
for i in range(1, 6):
    if i == 3:
        continue            # skip only this round
    print(i, end=" ")

print("\npass:")
for i in range(3):
    pass                    # legal do-nothing placeholder
print("pass loop ran silently")
````

## 13. Loop `else` (advanced)

## 14. `for` vs `while` — the decision

- Can you say "repeat ___ times" with a number? → `for`

**Predict:** what does this menu do when you run it? (Type 2 to escape!)

## 15. The 3-step pattern recipe + ONE worked demo

Recipe for every pattern:
1. **Outer loop = rows** (count rows in the picture)
2. **Inner loop(s) = one row's contents** — make a paper table: row number → spaces? stars? Find the formula.
3. **Newline after each row** — inner prints use `end=""`, then one empty `print()`.

**Worked demo: Hollow Square** (this is NOT one of your 8 — those you solve yourself):

```
* * * * *
*       *
*       *
*       *
* * * * *
```

Paper table: border rows (first/last) are all stars; middle rows are star + gaps + star.
Cell rule: star if first/last **row** OR first/last **column** — an `if` with `or`!

````python
n = 5
for row in range(1, n + 1):
    for col in range(1, n + 1):
        if row == 1 or row == n or col == 1 or col == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()   # close the row
````

Note the condition says `row == 1 or row == n` — NOT `row == 1 or n` (cell 5's trap, dodged).

Now change `n` to 8 and re-run. If your head predicted the new shape correctly, you own this recipe.
